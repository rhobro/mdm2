```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score
from imblearn.over_sampling import SMOTE
from scipy.spatial.distance import cdist
from scipy.stats import mode
from util import *

rand_state = rand_seed()
print(f"Using random state: {rand_state}")
```

```python
x_cols = ["age", "gender", "location", "size"]
y_col = "mdm2"
TEST_SIZE = 0.2

df = load_with_columns(x_cols + [y_col])

X = df.loc[:, x_cols]
Y = df[y_col].astype(bool)
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    X, Y,
    test_size=TEST_SIZE,
    random_state=rand_state,
    stratify=Y,
)
# x_train, y_train = SMOTE(
#     sampling_strategy=0.5,
#     random_state=rand_state,
# ).fit_resample(x_train, y_train)
```

```python
class KNN:
    def __init__(self, k):
        self.k = k
        self.x = None
        self.y = None
        self.genders = None
        self.locations = None

    def fit(self, x, y):
        x.gender, self.genders = x.gender.factorize()
        x.location, self.locations = x.location.factorize()
        self.x = x.values.astype(float)
        self.y = y.to_numpy()

    def predict(self, x):
        x.gender = self.genders.searchsorted(x.gender)
        x.location = self.locations.searchsorted(x.location)
        x = x.values.astype(float)

        dists = cdist(x, self.x, "euclidean")
        nearests = dists.argsort(axis=1)[:, :self.k]
        knn = self.y[nearests]
        # weights = 1 / np.take_along_axis(dists, nearests, axis=1)
        #
        # prob = (knn * weights).sum(axis=1) / weights.sum(axis=1)

        return knn.mean(axis=1)

        return knn.mean(axis=1)
```

```python
best_k = 0
best = 0

for i in range(1, 100):
    model = KNN(i)
    model.fit(x_train, y_train)
    y_hat = np.round(model.predict(x_test))
    score = accuracy_score(y_test, y_hat)

    if score > best:
        best = score
        best_k = i
    print(f"k = {i}, accuracy = {score}")

print(f"best: {best_k}, accuracy = {best}")
```
